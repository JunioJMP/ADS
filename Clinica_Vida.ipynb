```python
# === CLÍNICA VIDA+ CADASTRO ===

def cadastrar_paciente(pacientes):
    print("\n--- Cadastro de Paciente ---")
    nome = input("Nome do paciente: ").strip()
    if not nome:
        print("Nome não pode ser vazio.")
        return
    try:
        idade = int(input("Idade: "))
    except ValueError:
        print("Idade inválida. Digite um número inteiro.")
        return
    telefone = input("Telefone: ").strip()
    paciente = {
        "nome": nome,
        "idade": idade,
        "telefone": telefone
    }
    pacientes.append(paciente)
    print("Paciente cadastrado com sucesso!")


def ver_estatisticas(pacientes):
    print("\n--- Estatísticas ---")
    if not pacientes:
        print("Nenhum paciente cadastrado.")
        return
    total = len(pacientes)
    idades = [p["idade"] for p in pacientes]
    media = sum(idades) / total
    mais_novo = min(pacientes, key=lambda p: p["idade"])
    mais_velho = max(pacientes, key=lambda p: p["idade"])

    print(f"Total de pacientes: {total}")
    print(f"Idade média: {media:.1f} anos")
    print(f"Paciente mais novo: {mais_novo['nome']} ({mais_novo['idade']} anos)")
    print(f"Paciente mais velho: {mais_velho['nome']} ({mais_velho['idade']} anos)")


def buscar_paciente(pacientes):
    print("\n--- Buscar Paciente ---")
    nome_busca = input("Digite o nome do paciente: ").strip().lower()
    encontrados = [p for p in pacientes if nome_busca in p["nome"].lower()]
    if encontrados:
        print(f"\n {len(encontrados)} resultado(s) encontrado(s):")
        for p in encontrados:
            print(f"- {p['nome']} | {p['idade']} anos | Tel: {p['telefone']}")
    else:
        print("Nenhum paciente encontrado com esse nome.")


def listar_pacientes(pacientes):
    print("\n--- Lista de Pacientes ---")
    if not pacientes:
        print("Nenhum paciente cadastrado.")
        return
    for i, p in enumerate(pacientes, start=1):
        print(f"{i}. {p['nome']} - {p['idade']} anos - Tel: {p['telefone']}")


def menu():
    pacientes = []
    while True:
        print("\n=== CLÍNICA VIDA+ CADASTRO ===")
        print("1. Cadastrar paciente")
        print("2. Ver estatísticas")
        print("3. Buscar paciente")
        print("4. Listar todos os pacientes")
        print("5. Sair")
        opcao = input("Escolha uma opção: ").strip()
        if opcao == "1":
            cadastrar_paciente(pacientes)
        elif opcao == "2":
            ver_estatisticas(pacientes)
        elif opcao == "3":
            buscar_paciente(pacientes)
        elif opcao == "4":
            listar_pacientes(pacientes)
        elif opcao == "5":
            print("Saindo do sistema... Até logo!")
            break
        else:
            print("Opção inválida. Tente novamente.")

if __name__ == "__main__":
    menu()



    # === CONTROLE DE ATENDIMENTO - CLÍNICA VIDA+ ===

def verificar_consulta_normal(A, B, C, D):
    """Retorna se o paciente pode ser atendido em consulta normal."""
    return (A and B and C) or (B and C and D)


def verificar_emergencia(B, C, D):
    """Retorna se o paciente pode ser atendido em emergência."""
    return C and (B or D)


def entrada_logica(texto):
    """Lê uma resposta S/N e converte para valor lógico."""
    while True:
        resposta = input(texto + " (S/N): ").strip().upper()
        if resposta == "S":
            return True
        elif resposta == "N":
            return False
        else:
            print("Resposta inválida! Digite apenas S ou N.")


def menu():
    print("=== CONTROLE DE ATENDIMENTO - CLINICA VIDA+ ===")
    print("1. Consulta Normal")
    print("2. Emergência")
    print("3. Sair")
    while True:
        opcao = input("\nEscolha uma opção: ").strip()
        if opcao == "1":
            print("\n--- Verificação para CONSULTA NORMAL ---")
            A = entrada_logica("Paciente tem agendamento?")
            B = entrada_logica("Documentos estão em dia?")
            C = entrada_logica("Há médico disponível?")
            D = entrada_logica("Pagamentos estão em dia?")
            if verificar_consulta_normal(A, B, C, D):
                print("Atendimento LIBERADO para consulta normal!")
            else:
                print("Atendimento NEGADO. Verifique as condições.")
        elif opcao == "2":
            print("\n--- Verificação para EMERGÊNCIA ---")
            B = entrada_logica("Documentos estão em dia?")
            C = entrada_logica("Há médico disponível?")
            D = entrada_logica("Pagamentos estão em dia?")
            if verificar_emergencia(B, C, D):
                print("Atendimento de EMERGÊNCIA LIBERADO!")
            else:
                print("Atendimento NEGADO. Sem condições mínimas.")
        elif opcao == "3":
            print("Saindo do sistema...")
            break
        else:
            print("Opção inválida. Escolha 1, 2 ou 3.")


if __name__ == "__main__":
    menu()

```
